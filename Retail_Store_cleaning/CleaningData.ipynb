```python
complete=data.dropna(subset=["Price Per Unit","Quantity","Total Spent"])
check=(complete["Price Per Unit"]*complete["Quantity"]).round(2)
mismatch=(check!=complete["Total Spent"]).sum()
```

```python
print(f"mismatches out of {len(complete)} complete rows : {mismatch}") # we r tryna validate our data if the total spent is equal to the price per unit * Quantity 
```

```python
data[["Price Per Unit","Quantity","Total Spent"]].isna().sum() # this is gonna count the number of null values 
```

```python
trio=data[["Price Per Unit","Quantity","Total Spent"]] 
missing_per_row=trio.isna().sum(axis=1)# it calculates the num of missing values inside these 3 columns
missing_per_row.value_counts().sort_index() # it calculates the frequency of null values n sort their index as output
```

```python
one_missing=trio.isna()[missing_per_row == 1] # we select rows containing exactly one missing value
one_missing.idxmax(axis=1).value_counts() # Returns the column name containing True in each selected row & Counts how many times each column name appears.
```

```python
data
```

```python
two_missing=trio.isna()[missing_per_row==2] # we filtered the missing n got the name of column that has at least 2 missing values
two_missing.sum()# Counts the missing values (True) in each column
```

```python
mask=data["Price Per Unit"].isna() & data["Quantity"].notna() & data["Total Spent"].notna()  # we r looking for the row that has a null value on price per unit column , not having not null Quantity and not null on total spent 
```

```python
mask
```

```python
mask.value_counts() # 609 matched the condition 
```

```python
data.loc[mask,"Price Per Unit"]=data.loc[mask,"Total Spent"]/data.loc[mask,"Quantity"] # we r gonna select the mask that is matching the condition n calculate the price per unit \
    # price per unit=total spent/quantity
    # we calculate the price per unit by dividin its total spent and its quantity we filled the missing values w its correct price
    # loc allows us to get a specific rows with using .mask and a specific column 
```

```python
data[["Price Per Unit","Quantity","Total Spent"]].isna().sum() # now we r having 0 price per unit n filled it w the correct values
```

```python
data["Category"].value_counts()
```

```python
still_missing=data["Quantity"].isna() & data["Total Spent"].isna()
```

```python
still_missing.sum()
```

```python
data["Quantity_Estimated"]=still_missing
```

```python
data
```

```python
median_qty=data.groupby("Category")["Quantity"].median() # the median is less effected by the outliers 
```

```python
median_qty
```

```python
data.loc[still_missing,"Quantity"]=data.loc[still_missing,"Category"].map(median_qty) # this filled the missing values w the median of quantity grouped by the category 
```

```python
data.loc[still_missing,"Total Spent"]=data.loc[still_missing,"Price Per Unit"] * data.loc[still_missing,"Quantity"]
# after we mapped our values now we r gonna calculate the Total spent by multiplying the price per unit with the quantity 
```

```python
data[["Price Per Unit","Quantity","Total Spent"]].isna().sum() # now we r having zeros as null values 
```

```python
data
```

```python
data.groupby("Category")["Item"].apply(lambda x:x.isna().mean()).round(3) # now we calculated the percentage of the null vales at the Item column grouped by Category
```

```python
data.groupby("Location")["Item"].apply(lambda x:x.isna().mean()).round(3) # now we calculated the percentage of the null vales at the Item column grouped by Location
```

```python
data["Item"]=data["Item"].fillna("Unknown Item") # we labeled the missed values w unknown Item
data["Item"].isna().sum() # now we dont have missing values 
```

```python
data["Item"].value_counts()
```

```python
data.isna().sum()
```

```python
data.groupby(data["Transaction Date"].dt.year)["Discount Applied"].apply(lambda x:x.isna().mean()).round(3) # Calculates the percentage of missing values in "Discount Applied" for each year
```

```python
data.groupby(data["Location"])["Discount Applied"].apply(lambda x:x.isna().mean()).round(3) # Calculates the percentage of missing values in "Discount Applied" for each location
```

```python
data.groupby(data["Payment Method"])["Discount Applied"].apply(lambda x:x.isna().mean()).round(3) # Calculates the percentage of missing values in "Discount Applied" for each payment method
```

```python
data["Discount Applied"]=data["Discount Applied"].map({True:"Yes ", False:"No"}).fillna("Unknown") # convert every True with Yes and every False with No otherwise convert it to Unknown if it was a null value
```
