```python
data=data.drop(columns="rating")
```

```python
new_col_name=[]
for i in data.columns:
    new_col_name.append(i.capitalize())
```

```python
data.columns=new_col_name
```

```python
data.columns
```

```python
data.duplicated().sum()  # we dont have duplicates
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.isna().sum() # we dont have null vlaues
```

```python
# if we had any null values
data.dropna(inplace=True)
```

```python
data.head()
```

```python
data["Show_id"].unique() 
```

```python
data["Show_id"]=data["Show_id"].apply(lambda x: x.split("s")[1]) # iam gonna remove the s from the show_id column and return the only number.
```

```python
data[["Show_id"]]
```

```python
data["Show_id"]=data["Show_id"].astype(int)
```

```python
data["Date_added"]=data["Date_added"].apply(lambda x:x.replace("/","-"))
```
